#### fixing the column names

```python
df.columns = df.columns.str.lower().str.replace(' ','_')
df.columns
```

#### fixing the data types

```python
df['order_date'] = pd.to_datetime(df['order_date'],format='mixed')
```

```python
df['price_each'] = df['price_each'].astype(float)
df['quantity_ordered'] = df['quantity_ordered'].astype(int)
```

```python
df.info()
```

#### calculating some columns

```python
df['city'] = df['purchase_address'].str.split(',').str[1].str.strip()
```

```python
df['state'] = (df['purchase_address'].str.split(',').str[2].str.strip().str.split().str[0])
```

```python
df['total_amount'] = df['quantity_ordered'] * df['price_each']
```

```python
df['day_of_week'] = df['order_date'].dt.day_name()
```

```python
df['hour'] = df['order_date'].dt.hour
```

```python
df['order_day'] = df['order_date'].dt.date
```
